```python
# [CELL 1] My Inputs

STUDENT_ID = "TUPM-26-1838"      
LAST_NAME = "CULATA"       
FAVORITE_ARTIST = "ARIANA GRANDE" 

# AUTOMATED DATA PRE-PROCESSING
import re

# 1. Clean strings (remove spaces and special characters)
clean_name = re.sub(r'[^A-Z]', '', LAST_NAME.upper())
clean_artist = re.sub(r'[^A-Z]', '', FAVORITE_ARTIST.upper())

# 2. Extract values based on Appendix Rules
LAST_NAME_LENGTH = len(clean_name)
ARTIST_LENGTH = len(clean_artist)
SEED_DIGIT = int(STUDENT_ID[-1]) if STUDENT_ID else 0

# 3. Calculate numerical sums
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
ARTIST_SUM = sum((ord(char) - ord('A') + 1) for char in clean_artist)

print("--- STUDENT-SPECIFIC VALUES GENERATED ---")
print(f"SEED_DIGIT:    {SEED_DIGIT}")
print(f"ID_SUM:        {ID_SUM}")
print(f"NAME_LENGTH:   {LAST_NAME_LENGTH}")
print(f"ARTIST_LENGTH: {ARTIST_LENGTH}")
print(f"ARTIST_SUM:    {ARTIST_SUM}")
```

```python
# [CELL 2] EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH

def linear_search_ex1(records, target):
    comparisons = 0
    for i in range(len(records)):
        comparisons += 1
        if records[i] == target:
            return i, comparisons
    return -1, comparisons

def rotated_binary_search(records, target):
    low, high = 0, len(records) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        
        if records[mid] == target:
            return mid, comparisons
        
        # Check if the left half is normally sorted
        if records[low] <= records[mid]:
            if records[low] <= target < records[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Otherwise, the right half must be sorted
        else:
            if records[mid] < target <= records[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

# --- DATA GENERATION FOR EX1 ---
# Using student metrics to create a dynamic baseline dataset
base_start = ID_SUM * 10
original_ex1 = list(range(base_start, base_start + 100, 5))

# Rotation point determined strictly by your SEED_DIGIT
rotation_point = (SEED_DIGIT % len(original_ex1)) if SEED_DIGIT > 0 else 4
rotated_ex1 = original_ex1[rotation_point:] + original_ex1[:rotation_point]

# Target element picked reliably from within your dataset
target_index = len(original_ex1) // 2
target_incident = original_ex1[target_index]

# Run searches
lin_pos, lin_comps = linear_search_ex1(rotated_ex1, target_incident)
bin_pos, bin_comps = rotated_binary_search(rotated_ex1, target_incident)

print("=== EXERCISE 1: ASSESSMENT DATA ===")
print(f"Original Dataset: {original_ex1[:10]}... (Total {len(original_ex1)} elements)")
print(f"Rotated Dataset:  {rotated_ex1[:10]}...")
print(f"Rotation Point:   Index {rotation_point}")
print(f"Target Incident:  {target_incident}\n")
print(f"Linear Search Comparison Count: {lin_comps}")
print(f"Binary Search Comparison Count: {bin_comps}")
print(f"Final Output (Found at Position): {bin_pos}")
```

```python
# [CELL 3] EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS

def merge_and_track_ex2(dataset_a, dataset_b):
    consolidated = []
    i, j = 0, 0
    comparisons = 0
    
    while i < len(dataset_a) and j < len(dataset_b):
        comparisons += 1
        if dataset_a[i] <= dataset_b[j]:
            consolidated.append(dataset_a[i])
            i += 1
        else:
            consolidated.append(dataset_b[j])
            j += 1
            
    while i < len(dataset_a):
        consolidated.append(dataset_a[i])
        i += 1
        
    while j < len(dataset_b):
        consolidated.append(dataset_b[j])
        j += 1
        
    return consolidated, comparisons

# --- DATA GENERATION FOR EX2 ---
# Dataset A generates values using SEED_DIGIT and ID_SUM
start_a = (SEED_DIGIT + 1) * 10
dataset_A = list(range(start_a, start_a + (ID_SUM % 10 + 5) * 2, 2))

# Dataset B generates values using ARTIST_LENGTH and NAME_LENGTH
start_b = (ARTIST_LENGTH + 1) * 8
dataset_B = list(range(start_b, start_b + (LAST_NAME_LENGTH % 10 + 5) * 3, 3))

# Run the merge operation
consolidated_list, merge_comps = merge_and_track_ex2(dataset_A, dataset_B)
records_before = len(dataset_A) + len(dataset_B)
records_after = len(consolidated_list)
is_verified = consolidated_list == sorted(consolidated_list)

print("=== EXERCISE 2: ASSESSMENT DATA ===")
print(f"Department A Dataset: {dataset_A}")
print(f"Department B Dataset: {dataset_B}\n")
print(f"Consolidated Dataset: {consolidated_list}")
print(f"Number of Records Before Consolidation: {records_before}")
print(f"Number of Records After Consolidation:  {records_after}")
print(f"Comparison Count:    {merge_comps}")
print(f"Ordering Verification: {'PASSED (Sorted)' if is_verified else 'FAILED'}")
print(f"Final Output:         {consolidated_list}")
```

```python
# [CELL 4] EXERCISE 3: E-COMMERCE ORDER PROCESSING

import time

def bubble_sort_ex3(arr):
    a = arr.copy()
    n = len(a)
    comparisons, movements = 0, 0
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, movements

def selection_sort_ex3(arr):
    a = arr.copy()
    n = len(a)
    comparisons, movements = 0, 0
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            movements += 1
    return a, comparisons, movements

def insertion_sort_ex3(arr):
    a = arr.copy()
    n = len(a)
    comparisons, movements = 0, 0
    for i in range(1, n):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]
                movements += 1
                j -= 1
            else:
                break
        a[j + 1] = key
    return a, comparisons, movements

# --- DATA GENERATION FOR EX3 ---
# Construct a deterministic array length tied to your unique metadata metrics
base_val = (ID_SUM + ARTIST_SUM) % 20 + 10
# Scramble sequentially to build a predictable random-state pattern
random_input = [(x * 7 + 13) % 100 for x in range(base_val)]
sorted_input = sorted(random_input)
reverse_input = sorted_input[::-1]

inputs = {"Sorted": sorted_input, "Reverse": reverse_input, "Random": random_input}
algos = {"Bubble Sort": bubble_sort_ex3, "Selection Sort": selection_sort_ex3, "Insertion Sort": insertion_sort_ex3}

print("=== EXERCISE 3: ASSESSMENT DATA ===\n")
print(f"Generated Base Order Dataset: {random_input}\n")

for algo_name, algo_func in algos.items():
    print(f" {algo_name}:")
    for condition, data in inputs.items():
        start = time.perf_counter()
        res, comps, moves = algo_func(data)
        end = time.perf_counter()
        exec_ms = (end - start) * 1000
        
        print(f"  [{condition} Condition] -> Comparisons: {comps} | Movements: {moves} | Time: {exec_ms:.4f}ms")
    print("-" * 60)
```
